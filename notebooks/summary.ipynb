```python
import os
from pathlib import Path

import pandas as pd

# Resolve data directories from the environment variables run-notebooks injects.
output_dir = Path(os.environ.get("OUTPUT_DATA_DIR", "../output_data")).resolve()
source_dir = Path(os.environ.get("SOURCE_DATA_DIR", "../source_data")).resolve()
figures_base = Path(os.environ.get("FIGURES_DIR", output_dir / "figures")).resolve()
input_file = source_dir / "YNIMG_BrainParcellation_summary.tsv"

# Each notebook writes into its own folder, named after itself, under the
# shared figures tree. That folder is also what `run-notebooks` treats as the
# "already ran" marker: if it exists, the notebook is skipped. Writing
# somewhere else means the notebook re-runs on every `invoke run`.
summary_dir = figures_base / "summary"
summary_dir.mkdir(parents=True, exist_ok=True)
output_file = summary_dir / "authors.csv"

# Load the dataset
df = pd.read_csv(input_file, sep="\t")

# Extract authors (if column exists)
if "Author" in df.columns:
    authors = df[["Author"]]
    authors.to_csv(output_file, index=False)
    print(f"✅ Saved authors list to {output_file}")
else:
    print("❌ No 'Author' column found in the input data.")
```
